## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

url = "https://github.com/syedamominapak-coder/flyrankai_ml_internship/raw/refs/heads/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

declining_pages = (df["trend_direction"] == "down").sum()
declining_pct = declining_pages / len(df) * 100

print("Number of pages:", len(df))
print("Number of columns:", len(df.columns))
print("Pages with declining trend:", declining_pages)
print(f"Percentage of pages with declining trend: {declining_pct:.1f}%")

print("""
These numbers make the lane worth investigating because the dataset contains
30,000 pages and 44 columns, providing substantial page-level information.
16,262 pages have a declining trend, which is a large enough group to study
whether available page-level signals can help prioritize pages for review.
""")
```
